```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
f = open("../datasets/chat_data.txt", "r", encoding="utf-8")
data = f.read()
```

```python
print(data[:1000])  # Print the first 1000 characters of the chat data to get an overview
```

```python
print(type(data))
```

```python
pattern = '\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s-\s'
```

```python
import re
messages = re.split(pattern, data)[1:]
messages
```

```python
# extracting dates from the data
dates = re.findall(pattern,data)
dates
```

```python
# Create a DataFrame from the extracted messages
df = pd.DataFrame({'message_date': dates, 'user_message': messages})
df['message_date'] = pd.to_datetime(df['message_date'], format='%d/%m/%Y, %H:%M - ') 
```

```python
df.head()
```

```python
df.shape
```

```python
# seperating user and message
users = []
messages = []
for message in df['user_message']:
    entry = re.split('([\w\W]+?):\s',message)
    if entry[1:]:  # user name
        users.append(entry[1])
        messages.append(entry[2])
    else:
        users.append('group_notification')
        messages.append(entry[0])
df['user'] = users
df['message'] = messages
df.drop(columns=['message'], inplace=True)
df.head(20)    
```

```python
# extracting the dates from the chat data
df['year'] = df['message_date'].dt.year
```

```python
df.head()
```

```python
# extracting month name
df['month_num'] = df['message_date'].dt.month_name()
```

```python
# extracting dates from the chat data
df['day'] = df['message_date'].dt.day
```

```python
# extracting hours
df['hour'] = df['message_date'].dt.hour
```

```python
#extracting minutes
df['minute'] = df['message_date'].dt.minute
```

```python
# removing message_date column as we have extracted all the information from it
df.drop(columns=['message_date'], inplace=True)
```
